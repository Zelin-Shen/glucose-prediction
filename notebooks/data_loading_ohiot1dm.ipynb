```python
import os
import logging
import numpy as np
import pandas as pd
import xml.etree.ElementTree as ET
from tqdm import tqdm
```

```python
# 配置日志记录器
logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(name)s - %(levelname)s - %(message)s",
    handlers=[logging.StreamHandler()],
)  # 仅控制台输出
logger = logging.getLogger("Record")
```

```python
class OhioT1DMDataLoader:
    def __init__(self, train_directory, test_directory=None):
        self.train_directory = train_directory
        self.test_directory = test_directory
        self.train_data = None
        self.test_data = None

    def load_all_data(self):
        print("Loading OhioT1DM Dataset...")
        print("=" * 50)

        # Load training data
        self.train_data = self._load_data_from_directory(
            self.train_directory, "Training"
        )

        # Load testing data
        if self.test_directory:
            self.test_data = self._load_data_from_directory(
                self.test_directory, "Testing"
            )

        # Print summary
        print("\nData loading complete!")
        print(
            f"Training data: {len(self.train_data):,} records, {self.train_data['patient_id'].nunique()} patients"
        )
        if self.test_data is not None:
            print(
                f"Testing data: {len(self.test_data):,} records, {self.test_data['patient_id'].nunique()} patients"
            )

        return {"train": self.train_data, "test": self.test_data}

    def _load_data_from_directory(self, directory, data_type):
        print(f"\nLoading {data_type} Data:")
        print("-" * 30)

        # Find all files
        xml_files = [f for f in os.listdir(directory) if f.endswith(".xml")]

        if not xml_files:
            raise ValueError(f"No XML files found in {directory}")

        print(f"Found {len(xml_files)} XML files in {data_type.lower()} directory")

        # Process each file
        all_patient_data = []

        for xml_file in tqdm(xml_files, desc=f"Processing {data_type.lower()}"):
            file_path = os.path.join(directory, xml_file)
            patient_data = self._parse_patient_xml(file_path)

            if patient_data is not None and len(patient_data) > 0:
                # Add data split indicator
                patient_data["data_split"] = data_type.lower()
                all_patient_data.append(patient_data)
                logger.info(f"Loaded {len(patient_data)} records from {xml_file}")
            else:
                logger.warning(f"No valid data found in {xml_file}")

        if not all_patient_data:
            raise ValueError(
                f"No valid data could be loaded from {data_type.lower()} files"
            )

        # Combine all patient data
        combined_data = pd.concat(all_patient_data, ignore_index=True)

        print(
            f"{data_type} data: {len(combined_data):,} records, {combined_data['patient_id'].nunique()} patients"
        )
        print(
            f"Date range: {combined_data['timestamp'].min()} to {combined_data['timestamp'].max()}"
        )

        return combined_data

    def _parse_patient_xml(self, file_path):
        try:
            tree = ET.parse(file_path)
            root = tree.getroot()

            # extract patient ID
            patient_id = root.attrib.get(
                "id", os.path.basename(file_path).split("-")[0]
            )

            # extract important events
            glucose_events = self._extract_glucose_events(root)
            insulin_events = self._extract_insulin_events(root)
            meal_events = self._extract_meal_events(root)

            logger.debug(
                f"Patient {patient_id}: {len(glucose_events)} glucose, "
                f"{len(insulin_events)} insulin, {len(meal_events)} meal events"
            )

            if not glucose_events:
                logger.warning(f"No glucose events found for patient {patient_id}")
                return None

            # create time grid based on glucose timestamps
            df = self._create_time_grid(glucose_events, patient_id)

            # merge all event types
            df = self._merge_events(df, glucose_events, insulin_events, meal_events)

            return df

        except Exception as e:
            logger.error(f"Error parsing {file_path}: {str(e)}")
            return None

    def _extract_glucose_events(self, root):
        events = []

        for glucose_section in root.iter("glucose_level"):
            for event in glucose_section.findall("event"):
                timestamp_str = event.attrib.get("ts")
                value_str = event.attrib.get("value")

                if timestamp_str and value_str:
                    try:
                        timestamp = self._parse_timestamp(timestamp_str)
                        value = float(value_str)

                        # validation
                        if 20 <= value <= 400 and timestamp is not None:
                            events.append({"timestamp": timestamp, "glucose": value})
                    except (ValueError, TypeError):
                        continue

        return events

    def _extract_insulin_events(self, root):
        events = []

        # extract basal insulin
        for basal_section in root.iter("basal"):
            for event in basal_section.findall("event"):
                timestamp_str = event.attrib.get("ts")
                value_str = event.attrib.get("value")

                if timestamp_str and value_str:
                    try:
                        timestamp = self._parse_timestamp(timestamp_str)
                        value = float(value_str)

                        if 0 <= value <= 50 and timestamp is not None:
                            events.append(
                                {
                                    "timestamp": timestamp,
                                    "insulin": value,
                                    "insulin_type": "basal",
                                }
                            )
                    except (ValueError, TypeError):
                        continue

        # extract bolus insulin
        for bolus_section in root.iter("bolus"):
            for event in bolus_section.findall("event"):
                timestamp_str = event.attrib.get("ts_begin")
                dose_str = event.attrib.get("dose")

                if timestamp_str and dose_str:
                    try:
                        timestamp = self._parse_timestamp(timestamp_str)
                        dose = float(dose_str)

                        if 0 < dose <= 50 and timestamp is not None:
                            events.append(
                                {
                                    "timestamp": timestamp,
                                    "insulin": dose,
                                    "insulin_type": "bolus",
                                }
                            )
                    except (ValueError, TypeError):
                        continue

        return events

    def _extract_meal_events(self, root):
        events = []

        for meal_section in root.iter("meal"):
            for event in meal_section.findall("event"):
                timestamp_str = event.attrib.get("ts")
                carbs_str = event.attrib.get("carbs")

                if timestamp_str and carbs_str:
                    try:
                        timestamp = self._parse_timestamp(timestamp_str)
                        carbs = float(carbs_str)

                        if carbs > 0 and carbs <= 200 and timestamp is not None:
                            events.append({"timestamp": timestamp, "meal_carbs": carbs})
                    except (ValueError, TypeError):
                        continue

        return events

    def _parse_timestamp(self, timestamp_str):
        formats = ["%d-%m-%Y %H:%M:%S", "%Y-%m-%d %H:%M:%S"]

        for fmt in formats:
            try:
                timestamp = pd.to_datetime(timestamp_str, format=fmt)
                # for alignment
                return timestamp.round("5min")
            except ValueError:
                continue

        return None

    def _create_time_grid(self, glucose_events, patient_id):
        if not glucose_events:
            return pd.DataFrame()

        timestamps = [event["timestamp"] for event in glucose_events]
        start_time = min(timestamps)
        end_time = max(timestamps)

        # Create 5-minute intervals
        time_grid = pd.date_range(start=start_time, end=end_time, freq="5min")

        df = pd.DataFrame({"timestamp": time_grid, "patient_id": patient_id})

        return df

    def _merge_events(self, df, glucose_events, insulin_events, meal_events):
        # merge glucose events
        if glucose_events:
            glucose_df = pd.DataFrame(glucose_events)
            # aggregate multiple glucose readings at same timestamp
            glucose_df = glucose_df.groupby("timestamp")["glucose"].mean().reset_index()
            df = df.merge(glucose_df, on="timestamp", how="left")
        else:
            df["glucose"] = np.nan

        # merge insulin events
        if insulin_events:
            insulin_df = pd.DataFrame(insulin_events)
            # Sum insulin doses at same timestamp
            insulin_df = insulin_df.groupby("timestamp")["insulin"].sum().reset_index()
            df = df.merge(insulin_df, on="timestamp", how="left")
        else:
            df["insulin"] = np.nan

        # merge meal events
        if meal_events:
            meal_df = pd.DataFrame(meal_events)
            # Sum carbs at same timestamp
            meal_df = meal_df.groupby("timestamp")["meal_carbs"].sum().reset_index()
            df = df.merge(meal_df, on="timestamp", how="left")
        else:
            df["meal_carbs"] = np.nan

        # Fill NaN
        df["insulin"] = df["insulin"].fillna(0)  # No insulin = 0
        df["meal_carbs"] = df["meal_carbs"].fillna(0)  # No meal = 0
        # Keep glucose NaN for now (will handle in preprocessing)

        return df

    def get_basic_statistics(self):
        print("\nBasic Data Statistics:")
        print("=" * 50)

        # Training data statistics
        if self.train_data is not None:
            print("\nTRAINING DATA:")
            print("-" * 20)
            self._print_data_stats(self.train_data)

        # Testing data statistics
        if self.test_data is not None:
            print("\nTESTING DATA:")
            print("-" * 20)
            self._print_data_stats(self.test_data)

        # comparison
        if self.train_data is not None and self.test_data is not None:
            print("\nCOMPARISON:")
            print("-" * 15)
            train_patients = set(self.train_data["patient_id"].unique())
            test_patients = set(self.test_data["patient_id"].unique())

            print(f"Patients in both datasets: {train_patients == test_patients}")
            print(f"Training patients: {sorted(train_patients)}")
            print(f"Testing patients: {sorted(test_patients)}")

            if train_patients == test_patients:
                print("\n Same patients in both training and testing sets")
            else:
                print("\n Different patients:")
                print(f"  Only in training: {train_patients - test_patients}")
                print(f"  Only in testing: {test_patients - train_patients}")

    def _print_data_stats(self, data):
        print(f"Total records: {len(data):,}")
        print(f"Number of patients: {data['patient_id'].nunique()}")
        print(f"Time range: {data['timestamp'].min()} to {data['timestamp'].max()}")

        # Data completeness
        print("\nData Completeness:")
        print(
            f"Glucose readings: {data['glucose'].notna().sum():,} / {len(data):,} "
            f"({data['glucose'].notna().mean() * 100:.1f}%)"
        )
        print(f"Insulin events: {(data['insulin'] > 0).sum():,}")
        print(f"Meal events: {(data['meal_carbs'] > 0).sum():,}")

        # daily event frequencies
        print("\nDaily Event Frequencies:")
        daily_stats = []

        for patient_id in data["patient_id"].unique():
            patient_data = data[data["patient_id"] == patient_id]

            # Calculate actual days from timestamp range
            start_date = patient_data["timestamp"].min().date()
            end_date = patient_data["timestamp"].max().date()
            total_days = (
                end_date - start_date
            ).days + 1  # +1 to include both start and end days

            # Count events
            insulin_events = (patient_data["insulin"] > 0).sum()
            meal_events = (patient_data["meal_carbs"] > 0).sum()

            # Calculate daily rates
            insulin_per_day = insulin_events / total_days if total_days > 0 else 0
            meals_per_day = meal_events / total_days if total_days > 0 else 0

            daily_stats.append(
                {
                    "Patient_ID": patient_id,
                    "Total_Days": total_days,
                    "Insulin_Events": insulin_events,
                    "Meal_Events": meal_events,
                    "Insulin_per_Day": round(insulin_per_day, 2),
                    "Meals_per_Day": round(meals_per_day, 2),
                }
            )

        # Display daily frequency stats
        daily_df = pd.DataFrame(daily_stats)
        print(daily_df.to_string(index=False))

        # Overall averages
        total_days_all = daily_df["Total_Days"].sum()
        total_insulin_all = daily_df["Insulin_Events"].sum()
        total_meals_all = daily_df["Meal_Events"].sum()

        avg_insulin_per_day = (
            total_insulin_all / total_days_all if total_days_all > 0 else 0
        )
        avg_meals_per_day = (
            total_meals_all / total_days_all if total_days_all > 0 else 0
        )

        print("\nOverall Averages:")
        print(f"Insulin events per day: {avg_insulin_per_day:.2f}")
        print(f"Meal events per day: {avg_meals_per_day:.2f}")

        # Per-patient summary
        print("\nPer-Patient Summary:")
        patient_stats = (
            data.groupby("patient_id")
            .agg(
                {
                    "timestamp": ["count", "min", "max"],
                    "glucose": lambda x: x.notna().sum(),
                    "insulin": lambda x: (x > 0).sum(),
                    "meal_carbs": lambda x: (x > 0).sum(),
                }
            )
            .round(2)
        )

        patient_stats.columns = [
            "Total_Records",
            "Start_Date",
            "End_Date",
            "Glucose_Readings",
            "Insulin_Events",
            "Meal_Events",
        ]
        print(patient_stats)
```
