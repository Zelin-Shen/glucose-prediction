This is a simulated dataset describing type 1 diabetic patient time series data. It contains 4 channels:

- Glucose (collected by continuous glucose monitor, 288 samples per day),
- Meal (manually input by patients, normally 3-5 samples per day),
- Insulin (manually input by patients, normally 3-5 samples per day), and
- Time of the day.

The task is to predict the blood glucose level 30 mins (6 samples) later, given the previous G, M, I and time data.

The attribute information:

- blood glucose level: mg/dL
- insulin injected: unit
- meal intake: g

```python
### Import required libraries

import pandas as pd
import matplotlib.pyplot as plt
import scipy.io as sio
import numpy as np
import os

import random

seed = 1

import torch

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)
torch.cuda.manual_seed(seed)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False
os.environ["PYTHONHASHSEED"] = str(seed)

import warnings

warnings.filterwarnings("ignore")
```

```python
file_index = 1
data1 = sio.loadmat(
    "Data/simulated data/" + "adult360_" + str(file_index) + ".mat"
)  # load data from .mat files
data_history = data1["history"]

normalize_para = 100
Glucose = data_history[0]["CGM"][0] / normalize_para  # 100 mg/dL

BW = 70
Insulin = data_history[0]["u6"][0] / 6000 * BW  # unit

Ts = 5
Meal = data_history[0]["u1"][0] / 1000 * Ts  # g
```

```python
Glucose.shape[0] / 288  # each day has 288 points, and there are 360 days
```

```python
day_index = np.array([x / 288 for x in range(288)] * 360)
day_index = day_index.reshape(len(day_index), 1)

data_full = np.concatenate((Glucose, Insulin, Meal, day_index), axis=1)
df_full = pd.DataFrame(data=data_full, columns=["G", "I", "M", "T"])
df_full.head(110)
```
